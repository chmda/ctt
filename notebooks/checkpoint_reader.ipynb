```python
import os
import pickle
import ipywidgets as widgets
import matplotlib.pyplot as plt
import jax.numpy as jnp
from IPython.display import display, clear_output
```

```python
FOLDER = os.path.join("..", "checkpoints")
file_list = list(filter(lambda f: os.path.isfile(os.path.join(FOLDER, f)), os.listdir(FOLDER)))

dropdown = widgets.Dropdown(
    options=file_list,
    description="Select file:",
    disabled=False
)
output = widgets.Output()

def on_file_selected(change):
    with output:
        clear_output(wait=True)
        plt.close("all")
        selected_file = change["new"]
        file_path = os.path.join(FOLDER, selected_file)

        splits = selected_file.split("-")
        target_function = splits[2]
        with open(file_path, "rb") as f:
            data = pickle.load(f)
            print(f"Target function: {target_function}")
            print(f"Dimension: {data['dim']}")
            print(f"Basis dimension: {data['n_features']}")
            print(f"Number of layers: {data['num_steps']}")
            print(f"Ranks: {data['ranks']}")
            print(f"Regularization coefficient: {data['reg_coeff']}")

            rel_l2 = data['rel_l2']
            plt.semilogy(
                jnp.arange(len(rel_l2)),
                rel_l2,
                label="Relative L2"
            )
            plt.grid()
            plt.legend()
            plt.xlabel("Iterations")
            plt.ylabel("Error")
            plt.show()

dropdown.observe(on_file_selected, names="value")
display(dropdown, output)
```
